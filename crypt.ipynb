```python
for A in range(1, 10):
    for B in range(1, 10):
        C = A + B

        if C < 10 and A != B and A != C and B != C:
            print("A =", A)
            print("B =", B)
            print("C =", C)
            print(A, "+", B, "=", C)
            exit()
```
